## Theory Card — Polymorphism + Encapsulation + Abstraction

POLYMORPHISM — same method name, different behaviour per class.
Real life: A remote control. The same PLAY button works on a TV, music player, and DVD player.
Same command. Different device. Different action.

ENCAPSULATION — hide internal data, control access through methods.
Real life: An ATM machine. You can deposit and withdraw but you cannot directly open the machine and change the balance database.
Single underscore _x means: please do not touch (convention, Python allows it).
Double underscore __x means: Python blocks direct access using name mangling.

ABSTRACTION — define what must exist, not how it works.
Real life: A job description. It says the role must have connect(), query(), close(). Each candidate implements differently.
Abstract class = job description. Concrete class = the actual employee.
You cannot create an object from an abstract class directly — it will throw TypeError.

ENCAPSULATION vs ABSTRACTION — most people confuse these:
Encapsulation = hiding the data (what is stored inside)
Abstraction = hiding the complexity (how something works)
ATM hides the database = encapsulation.
You do not need to know how the bank processes the transaction = abstraction.

IN DATA ENGINEERING:
Polymorphism — your pipeline.run() works whether the connector is MySQL, HDFS, or Kafka.
Encapsulation — your config class hides credentials, only exposes get_connection().
Abstraction — your abstract Connector forces every connector to implement connect() and query().

### Fix 4 — Q13 enumerate in correct order
You used reversed() which printed lines backwards. Write it in normal order.

```python
# Fix 4
with open("file.txt", "w") as f:
    f.write("line one\nline two\nline three\n")

with open("file.txt", "r") as f:
    for num, line in enumerate(f, start=1):
        print(num, line.strip())
```
