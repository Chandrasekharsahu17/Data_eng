## Theory Card — Polymorphism + Encapsulation + Abstraction

POLYMORPHISM — same method name, different behaviour per class.
Real life: A remote control. The same PLAY button works on a TV, music player, and DVD player.
Same command. Different device. Different action.

ENCAPSULATION — hide internal data, control access through methods.
Real life: An ATM machine. You can deposit and withdraw but you cannot directly open the machine and change the balance database.
Single underscore _x means: please do not touch (convention, Python allows it).
Double underscore __x means: Python blocks direct access using name mangling.

ABSTRACTION — define what must exist, not how it works.
Real life: A job description. It says the role must have connect(), query(), close(). Each candidate implements differently.
Abstract class = job description. Concrete class = the actual employee.
You cannot create an object from an abstract class directly — it will throw TypeError.

ENCAPSULATION vs ABSTRACTION — most people confuse these:
Encapsulation = hiding the data (what is stored inside)
Abstraction = hiding the complexity (how something works)
ATM hides the database = encapsulation.
You do not need to know how the bank processes the transaction = abstraction.

IN DATA ENGINEERING:
Polymorphism — your pipeline.run() works whether the connector is MySQL, HDFS, or Kafka.
Encapsulation — your config class hides credentials, only exposes get_connection().
Abstraction — your abstract Connector forces every connector to implement connect() and query().

### HW3 — Abstraction with DataConnector
Write abstract class DataConnector with abstract methods: connect(), query(sql), close().
Implement MySQLConnector and HDFSConnector — each prints different messages per method.
Write function run_pipeline(connector, sql) that calls all 3 methods in order.
Test with both connectors.
Show that DataConnector() alone raises TypeError.

```python
# HW3
from abc import ABC, abstractmethod
class DataConnector(ABC):
    @abstractmethod
    def connect(self):
        pass
    @abstractmethod
    def query(self,sql):
        pass
    @abstractmethod
    def close(self):
        pass

class HDFSConnector(DataConnector):
    def connect(self):
        return "HDFS connected"

    def query(self,sql):
        return"HDFS query has been implemnted"

    def close(self):
        return "HDFS closed"

class MySQLConnector(DataConnector):
    def connect(self):
        return "mys sql connected"

    def query(self,sql):
        return"query has been implemnted"

    def close(self):
        return "sql closed"

def run_pipeline(connector,sql):
    o=connector()
    o.connect()
    o.query(sql)
    o.close()

print("---- MySQL Pipeline ----")
run_pipeline(MySQLConnector, "SELECT * FROM users")
```

### HW4 — Property with validation
Create class Rectangle with private __width and __height.
@property for width and height that return values.
@width.setter and @height.setter that raise ValueError if value is negative or zero.
@property area — returns width times height.
@property perimeter — returns 2 times (width + height).
Test with valid values. Test with negative value to trigger the error.

```python
# HW4
class Rectangle():
    def __init__(self,width,height):
        self.__height=height
        self.__width=width
    @property 
    def width(self):
        return self.__width

    @width.setter
    def width(self,value):
        if value<0:
            raise ValueError("Value cannot be negative")
        self.__width=value
        
    @property 
    def height(self):
        return self.__height        
        
    @height.setter
    def height(self,value):
        if value<=0:
            raise ValueError("Value cannot be zero or negative")
        self.__height=value

    @property
    def Area(self):
        return self.__height*self.__width
        
    @property
    def Perimeter(self):
        return 2*(self.__height+self.__width)    

r = Rectangle(5, 10)

print("Width:", r.width)
print("Height:", r.height)
print("Area:", r.Area)
print("Perimeter:", r.Perimeter)
```

### HW5 — Mixed (All OOP pillars together)
Build a mini data pipeline using all three pillars.

Abstract class Pipeline:
  Abstract methods: extract(), transform(), load()
  Non-abstract method run() that calls all three in order and prints start and end time

Class SalesPipeline(Pipeline):
  extract() — prints: Reading from sales.csv
  transform() — prints: Cleaning nulls and fixing data types
  load() — prints: Writing to database

Class LogPipeline(Pipeline):
  extract() — prints: Reading server logs
  transform() — prints: Filtering ERROR lines only
  load() — prints: Writing to monitoring dashboard

Create both pipelines and call run() on each.
Notice how run() is the same for both but the steps are different — that is polymorphism + abstraction together.

```python
# HW5
from abc import ABC, abstractmethod
from datetime import datetime

class Pipeline(ABC):
    @abstractmethod 
    def extract(self):
        pass
    @abstractmethod 
    def transform(self):
        pass
    @abstractmethod 
    def load(self):
        pass
    def run(self):
        print("pipeline started")
        self.extract()
        self.transform()
        self.load()
        print("pipeline finished")
    
class SalesPipeline(Pipeline):
   
    def extract(self):
        print("read and analyse csvfile")

    def transform(self):
        print("transform CSV files")
  
    def load(self):
        print("load the csv files")
                        
class LogPipeline(Pipeline):
   
    def extract(self):
        print("read log files")

    def transform(self):
        print("transform log")
  
    def load(self):
        print("writing to moniotoer dashboard")

sales = SalesPipeline()
logs = LogPipeline()

sales.run()
print("-----")
logs.run()
```
