## 📘 Exception Handling — Key Concepts

```python
try:
    # code that might fail
except FileNotFoundError:
    # runs if that specific error occurs
except ValueError as e:
    # 'as e' lets you access the error message
    print(e)
else:
    # runs ONLY if NO exception occurred
finally:
    # ALWAYS runs — use for cleanup
```

**Rule:** Always catch the most specific error. Never use bare `except Exception` — it hides bugs.

**Custom exceptions:** Inherit from `Exception`:
```python
class MyError(Exception):
    pass

raise MyError("something went wrong")
```

## 📘 OOP Concepts — Read Before Watching

### The 4 Pillars
- **Encapsulation** — bundle data + functions together in a class, hide internal details
- **Inheritance** — child class gets all parent class methods/attributes
- **Polymorphism** — same method name, different behaviour depending on class
- **Abstraction** — expose only what's needed, hide complexity

### Class basics
```python
class Dog:
    species = 'Canis lupus'   # class attribute — shared by ALL dogs

    def __init__(self, name):  # runs when object is created
        self.name = name       # instance attribute — unique to THIS dog

    def bark(self):
        print(f'{self.name} says woof!')

d = Dog('Rex')
d.bark()  # Rex says woof!
```

### Inheritance
```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print('...')

class Dog(Animal):           # Dog inherits from Animal
    def speak(self):         # overrides parent method
        print(f'{self.name} says woof!')

class Cat(Animal):
    def speak(self):
        print(f'{self.name} says meow!')

# super() calls the parent class method
class GuideDog(Dog):
    def __init__(self, name, owner):
        super().__init__(name)   # calls Dog/__init__ → Animal.__init__
        self.owner = owner
```

### Why this matters in Big Data
Every real DE project uses OOP. Your pipeline stages (Extract, Transform, Load) are classes. Your connectors (HDFSConnector, S3Connector) share a base class. Configuration objects are classes. You can't avoid it.
